```python
from langchain.agents import create_agent
from langchain.tools import tool
import bs4


from dotenv import load_dotenv
from __future__ import annotations
import os

try:
    load_dotenv()
    API_KEY = os.getenv("GOOGLE_API_KEY")
except Exception as e:
    print(f"Error loading environment variable(s): {e}")
```

```python
@tool(description = "Get the current weather for a given city")
def get_weather(location: str) -> str:
    return f"The current weather in {location} is sunny with a temperature of 25°C."

agent = create_agent(
    model="google_genai:gemini-3.5-flash-lite",
    tools=[get_weather],
    system_prompt="You are a helpful assistant"
)

result = agent.invoke(
    {"messages": [{"role": "user", "content": "What's the weather in San Francisco?"}]}
)
print(result["messages"][-1].content_blocks)
```

```python
from google import genai
from google.genai import types

client = genai.Client(api_key=API_KEY)
result = client.models.embed_content(
        model="gemini-embedding-001",
        contents="What is the meaning of life?",
        config = types.EmbedContentConfig(task_type = "SEMANTIC_SIMILARITY", output_dimensionality = 1536)
)

print(result.embeddings)
```

```python
from langchain_core.prompts import ChatPromptTemplate
from langchain_google_genai import ChatGoogleGenerativeAI
from langchain_core.output_parsers import StrOutputParser
from langchain_core.runnables import RunnablePassthrough

template = """
Answer the question based on the context below:
{context}
Question: {question}.

In case the answer is not contained within the context, you have to mention it.
"""

prompt = ChatPromptTemplate.from_messages([
    ("user", template)
])

llm = ChatGoogleGenerativeAI(model = "gemini-3.5-flash", temperature = 0)
rag_llm_chain = (
    {"context": "The temperature is 35 degrees in Nantes.", "question" : RunnablePassthrough()}
    | prompt 
    | llm
    | StrOutputParser()
)

rag_llm_chain.invoke("Hello what is the temperature in Nantes ?")
```
